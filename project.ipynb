```python
import pandas as pd
data1 = pd.read_csv("players_data-2024_2025.csv")
data2 = pd.read_csv("players_data-2025_2026.csv")
```

```python
#data size
print ("size(2024-2025):",data1.size)
print ("size (2025-2026):", data2.size)
```

```python
import numpy as np
Age= data1["Age"]
mean = np.mean(Age)
maximum = np.max(Age)
minimum = np.min(Age)
standard = np.std(Age)

print("Mean Age:" ,mean)
print("Maximum Age:" ,maximum)
print("Minimum Age:", minimum)
print("Standard Deviation:", standard)
```
