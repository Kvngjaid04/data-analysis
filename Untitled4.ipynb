```python
user_data = input().split()

for s in user_data:
    count = 0
    for c in s.lower():
        if c in {"a", "e", "i", "o", "u"}:
            count += 1
    print(s + " " + str(count))
```

```python
def output_var_args(param1, *args):
    print(param1)
    for arg  in args:
        print(arg)

output_var_args("a", "b", "c")
```

```python
def output_var_kwargs(param1, **kwargs):
    print(param1)
    for key, arg in kwargs.items():
        print(key, arg)

output_var_kwargs("a", x="b", y="c")
```

```python
def is_palindrome(word):
    i= 0
    if len(word) < 2:
        return False
    if word[0] != word[-1]:
        return False
    while i in range((len(word)//2)):
        if word[0+i] != word[-1-i]:
             return False
        else :
            return True
        i += 1
```
